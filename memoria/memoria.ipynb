# Proyecto EDA: Optimización del Consumo Energético Industrial

## 1. Descripción del Proyecto

Este proyecto consiste en un Análisis Exploratorio de Datos (EDA) centrado en el consumo eléctrico de una planta siderúrgica.

El objetivo es entender cómo se comporta el consumo eléctrico en función del clima, los horarios de operación y los periodos tarifarios eléctricos en España. El análisis busca detectar patrones de consumo, posibles consumos no productivos y oportunidades de optimización energética y económica.

El consumo original se utiliza como patrón industrial y se adapta al calendario tarifario de análisis de 2024.

El coste calculado en este proyecto corresponde únicamente a una estimación del coste de energía activa. No representa una factura eléctrica completa, ya que no incluye potencia contratada, excesos de potencia, penalizaciones por reactiva, impuestos, IVA ni otros conceptos contractuales.

---

## 2. Objetivos Principales

* **Análisis climático:** Explorar si existen indicios de relación entre temperatura y consumo eléctrico agregado.
* **Análisis tarifario:** Evaluar la distribución del consumo en los periodos P1-P6 para identificar en qué tramos se concentra más energía y más coste.
* **Optimización económica:** Comparar el peso del consumo y el peso del coste por periodo tarifario, identificando posibles oportunidades de desplazamiento de consumo.
* **Detección de consumos base:** Identificar consumos relevantes en horarios de baja actividad que podrían indicar cargas permanentes, standby o consumos no productivos.
* **Análisis eléctrico complementario:** Revisar la energía reactiva y el factor de potencia como señales técnicas de posible ineficiencia eléctrica.

---

## 3. Datasets Utilizados

* **Consumo industrial:** *Steel Industry Energy Consumption* (UCI Machine Learning Repository). Datos de potencia activa, potencia reactiva, factor de potencia y tipo de carga con frecuencia de 15 minutos.
* **Contexto climático:** Histórico climático diario de 2024 correspondiente a la ubicación de la planta. Se utiliza para explorar la relación entre temperatura y consumo agregado.
* **Económico / tarifario:** Matriz de periodos horarios de la tarifa 3.0TD peninsular y precios por periodo P1-P6 para el calendario de análisis de 2024.
* **Tarifa completa:** Dataset con información adicional de tarifa eléctrica. Se reserva para una posible fase posterior de estimación más completa del coste eléctrico.

---

## 4. Preguntas de Investigación e Hipótesis

1. **H1:** El consumo eléctrico diario puede variar en función de la temperatura exterior, especialmente en días de frío o calor intenso.

2. **H2:** Una parte relevante del consumo puede concentrarse en periodos tarifarios caros, especialmente P1 y P2.

3. **H3:** El impacto económico del consumo no depende solo de los kWh consumidos, sino también del periodo tarifario en el que se producen.

4. **H4:** Existen consumos relevantes en horarios de baja actividad que podrían indicar cargas permanentes, standby o consumos no productivos.

---

## 5. Metodología Brief

El análisis se realizará mediante la integración de los datasets de consumo, clima y tarifa usando Python y Pandas.

Las principales tareas metodológicas serán:

* Limpieza y validación de fechas, consumos, periodos y precios.
* Agregación del consumo de 15 minutos a nivel horario para el análisis tarifario.
* Asignación de cada registro horario a su periodo tarifario P1-P6.
* Cálculo del coste estimado de energía activa mediante la fórmula:

  `coste_energia_activa = consumo_kwh × precio_eur_kwh`

* Análisis de patrones por hora, día de la semana, mes, tipo de día y periodo tarifario.
* Comparación entre porcentaje de consumo y porcentaje de coste por periodo.
* Identificación de posibles consumos base o anomalías en horarios de baja actividad.
* Exploración de la relación entre temperatura diaria y consumo diario agregado.
* Visualización de patrones mediante gráficos temporales, distribuciones, heatmaps y comparativas por periodo.

---

## 6. Alcance y Limitaciones

Este proyecto debe interpretarse como un EDA exploratorio y una simulación tarifaria, no como una reconstrucción exacta de una factura eléctrica real.

Las principales limitaciones son:

* El consumo original se adapta al calendario de análisis, por lo que las conclusiones deben centrarse en patrones y no en importes reales absolutos.
* El coste calculado solo incluye energía activa.
* No se dispone de datos reales de producción, turnos, paradas, órdenes de fabricación o mantenimiento.
* Las conclusiones sobre consumos base, clima o anomalías deben considerarse indicios exploratorios, no causas demostradas.
* Para validar oportunidades reales de ahorro sería necesario contrastar los hallazgos con responsables de producción, mantenimiento, energía o ingeniería.

---

## 7. Resultado Esperado

El resultado esperado es obtener una visión clara del comportamiento energético de la planta y responder a las siguientes preguntas:

* ¿Cuándo consume más electricidad la planta?
* ¿Cómo cambia el consumo según temperatura, hora, día y periodo tarifario?
* ¿Qué periodos concentran más energía y cuáles generan más coste?
* ¿Existe diferencia entre el peso energético y el peso económico de cada periodo?
* ¿Aparecen consumos relevantes en horarios de baja actividad?
* ¿Qué oportunidades podrían investigarse para reducir consumo, desplazar cargas o mejorar el coste energético?

El objetivo final es convertir los datos de consumo eléctrico en conclusiones prácticas para orientar decisiones de optimización energética industrial.
