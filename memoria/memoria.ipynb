# Proyecto EDA: Optimización del Consumo Energético Industrial

## 1. Descripción del Proyecto
Este proyecto consiste en un Análisis Exploratorio de Datos (EDA) centrado en una planta siderúrgica durante el año 2023. El objetivo es entender cómo las variables externas (clima) y las estructuras de costes (tarifas eléctricas en España) afectan la eficiencia operativa y el gasto económico de la industria.

## 2. Objetivos Principales
*   **Correlación Climática:** Determinar si existe un aumento del consumo eléctrico vinculado a temperaturas extremas (frío/calor). Asi como la afeccion de esto al tarifario electrico
*   **Análisis Tarifario:** Evaluar la distribución del consumo en los periodos P1-P6 de la tarifa 3.0TD para identificar periodos economicos mas optimos tanto por consumo como por coste
*   **Detección de Anomalías:** Identificar "consumos fantasma" en periodos de inactividad o paradas técnicas.

## 3. Datasets Utilizados
*   **Consumo (Micro):** *Steel Industry Energy Consumption* (UCI Machine Learning Repository). Datos de potencia activa y reactiva cada 15 minutos.
*   **Contexto (Clima):** Histórico de temperaturas diarias de 2023 (Fuente: AEMET/Visual Crossing).
*   **Económico (Tarifas):** Matriz de periodos horarios de la tarifa 3.0TD peninsular (España, 2023).

## 4. Preguntas de Investigación (Hipótesis)
1.  **H1:** El consumo eléctrico presenta picos significativos cuando la temperatura exterior supera los 25°C debido a sistemas de refrigeración.
2.  **H2:** Más del 20% del gasto energético ocurre en periodos Punta (P1-P2), a pesar de ser los tramos más caros.
3.  **H3:** El factor del cambio de turnos/horarios de trabajo afectarian en el coste energetico de la planta

## 5. Metodología Brief
El análisis se realizará mediante la integración de los tres datasets usando Python (Pandas). Se aplicarán técnicas de resampleo temporal, ingeniería de variables para los tramos horarios y visualizaciones avanzadas con Seaborn y Plotly para identificar patrones estacionales y diarios.
