# Proyecto EDA: Optimización del Consumo Energético Industrial

## 1. Descripción del Proyecto

Este proyecto consiste en un Análisis Exploratorio de Datos aplicado al consumo eléctrico de una instalación industrial.

El objetivo es entender cómo se comporta el consumo eléctrico según el calendario, los turnos de trabajo, los periodos tarifarios P1-P6 y determinadas variables climáticas. A partir de este análisis se busca identificar patrones de consumo y posibles oportunidades de optimización energética y económica.

El consumo original se utiliza como patrón industrial y se adapta al calendario de análisis de 2024. El análisis no pretende reconstruir una factura eléctrica real completa, sino estimar el impacto económico parcial del consumo según la estructura tarifaria disponible.

---

## 2. Objetivos Principales

Los objetivos principales del proyecto son:

* **Análisis climático:** explorar si existen indicios de relación entre temperatura, viento y consumo eléctrico agregado.
* **Análisis operativo:** analizar el consumo por hora, día, semana y turno de trabajo.
* **Análisis tarifario:** evaluar la distribución del consumo en los periodos P1-P6.
* **Optimización económica:** comparar el peso del consumo y el peso del coste estimado por periodo tarifario.
* **Simulación horaria:** analizar si un cambio en el horario de trabajo podría afectar al coste energético estimado.
* **Análisis eléctrico complementario:** revisar energía reactiva y factor de potencia como posibles señales técnicas de ineficiencia eléctrica.

---

## 3. Datasets Utilizados

Para el desarrollo del EDA se utilizan los siguientes datasets:

* **`Steel_industry_data.csv`**  
  Dataset base de consumo eléctrico industrial con registros cada 15 minutos. Incluye consumo activo, energía reactiva, factor de potencia, tipo de carga y variables temporales originales.

* **`periodos_peninsula.csv`**  
  Tabla de periodos tarifarios P1-P6 según mes, tipo de día y hora. Se utiliza para asignar a cada hora de consumo su periodo tarifario correspondiente.

* **`clima_aemet_espana_media_diaria_2024.csv`**  
  Dataset climático diario de 2024. Incluye temperatura media, mínima, máxima, precipitación, viento, horas de sol y otras variables meteorológicas.

* **`tarifa_completa_2024.csv`**  
  Dataset económico con información de tarifa eléctrica por mes y periodo. Incluye términos de potencia, energía regulada, excesos, reactiva, impuestos y otros conceptos.

A partir de estos archivos se generan datasets limpios e integrados. El dataset principal del análisis es:

`df_maestro.csv`

Este dataset integra consumo horario, calendario 2024, turno, periodo tarifario, tarifa y coste estimado parcial.

---

## 4. Preguntas de Investigación e Hipótesis

El análisis se organiza en torno a las siguientes hipótesis:

1. **H1:** El consumo eléctrico diario puede variar en función de la temperatura exterior, especialmente en días de frío o calor intenso.

2. **H2:** Una parte relevante del consumo puede concentrarse en periodos tarifarios caros, especialmente P1 y P2.

3. **H3:** El impacto económico del consumo no depende solo de los kWh consumidos, sino también del periodo tarifario en el que se producen.

4. **H4:** El cambio de turnos u horarios de trabajo puede afectar al coste energético estimado de la planta.

---

## 5. Metodología Brief

El análisis se realiza con Python, Pandas, Matplotlib y Seaborn.

Las principales tareas metodológicas son:

* Limpieza y preparación de los datasets originales.
* Adaptación del patrón de consumo al calendario 2024.
* Creación de variables temporales: fecha, hora, día, mes, semana, día de la semana y turno.
* Agregación del consumo desde registros de 15 minutos a datos horarios.
* Asignación de cada hora a un periodo tarifario P1-P6.
* Integración del consumo horario con la tarifa eléctrica.
* Cálculo de un coste estimado parcial.

El coste estimado parcial se calcula como:

`coste_estimado_parcial = consumo_kwh × precio_variable_parcial_eur_kwh`

Donde:

`precio_variable_parcial_eur_kwh = energia_regulada_eur_kwh + fnee_eur_kwh`

También se realiza una simulación de escenario alternativo adelantando dos horas el patrón de consumo para observar su posible impacto económico.

---

## 6. Alcance y Limitaciones

Este proyecto debe interpretarse como un EDA exploratorio y una simulación energética, no como una auditoría energética completa.

Las principales limitaciones son:

* El consumo original procede de un patrón industrial adaptado metodológicamente al año 2024.
* No se dispone de datos reales de producción, paradas, mantenimiento, órdenes de fabricación o turnos reales.
* El clima utilizado es diario y agregado, por lo que solo permite un análisis exploratorio.
* El coste calculado es parcial y no equivale a una factura eléctrica real.
* No se incluyen todos los conceptos contractuales de una factura eléctrica industrial.
* La simulación de adelantar dos horas el consumo no considera restricciones reales de proceso, personal, equipos o planificación productiva.

Por tanto, los resultados deben entenderse como una primera aproximación analítica para detectar patrones y oportunidades potenciales.

---

## 7. Resultado Esperado

El resultado esperado es disponer de una visión clara y estructurada del comportamiento energético de la planta.

El EDA debe permitir responder a preguntas como:

* ¿Cuándo consume más electricidad la planta?
* ¿Cómo se distribuye el consumo por turnos?
* ¿Cómo cambia el consumo según día, hora y periodo tarifario?
* ¿Qué periodos concentran más consumo?
* ¿Qué periodos tienen mayor peso económico estimado?
* ¿Existe diferencia entre el peso energético y el peso económico de cada periodo?
* ¿El clima muestra alguna relación exploratoria con el consumo o el coste estimado?
* ¿Un cambio horario del patrón de consumo podría modificar el coste estimado?

El objetivo final es transformar los datos de consumo eléctrico en información útil para interpretar el comportamiento energético industrial y orientar posibles decisiones de optimización.
